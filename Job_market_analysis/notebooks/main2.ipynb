```python
driver=webdriver.Chrome(service=Service(ChromeDriverManager().install()))
```

```python
#pagination
alljobs=[]
for page in range(2,11):
    url=f"https://www.naukri.com/data-scientist-jobs-{page}?k=data%20scientist&nignbevent_src=jobsearchDeskGNB"
    driver.get(url)
    print("="*50)
    print(f"Currently scraping Page {page}")
    print(driver.current_url)
    print(soup.title)
    time.sleep(3)
    print(driver.title)
    print(driver.page_source[:500])
    soup=BeautifulSoup(driver.page_source,'html.parser')
    container=soup.find('div',id='listContainer')
    jobs=container.find_all('div',class_='srp-jobtuple-wrapper')
    for job in jobs:
        try:
            title = job.find('a', class_='title').text.strip()
            comp = job.find('a', class_='comp-name mw-25').text.strip()
            rating = job.find('a', class_='rating').text.strip()
            exp = job.find('span', class_='exp-wrap').text.strip()
            loc = job.find('span', class_='loc-wrap ver-line').text.strip()
            description = job.find(
            'span',
            class_='job-desc ni-job-tuple-icon ni-job-tuple-icon-srp-description'
            ).text.strip()

            skills = job.find_all('li', class_='dot-gt tag-li')
            skills_tag = [skill.text.strip() for skill in skills]

            alljobs.append({
            "title": title,
            "company": comp,
            "rating": rating,
            "experience": exp,
            "location": loc,
            "description": description,
            "skills": skills_tag
            })

        except Exception:
            continue
```

```python
import pandas as pd
df=pd.DataFrame(alljobs)
```

```python
data["min_experience"] = data["experience"].str.extract(r"(\d+)-")[0].astype(float)
data["max_experience"] = data["experience"].str.extract(r"-(\d+)")[0].astype(float)
```

```python
import numpy as np
data.loc[data["experience"] == "08 Aug", ["min_experience", "max_experience", "avg_experience"]] = np.nan
```

```python
def categorize_experience(x):
    if pd.isna(x):
        return np.nan
    elif x <= 2.5:
        return "Beginner"
    elif x <= 6:
        return "Intermediate"
    else:
        return "Advanced"
```

```python
data["experience_level"] = data["avg_experience"].apply(categorize_experience)
```

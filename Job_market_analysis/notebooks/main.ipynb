service module-
    from selenium.webdriver.chrome.service import Service 
        Service helps Selenium connect to the ChromeDriver executable.

Python ↔ ChromeDriver ↔ Chrome browser

Service manages that middle bridge.(without it python can't talk with chrome properly.)

chromedrivermanager-as chrome itself can not be controller directly. so it needs a driver.

so the driver is: chromedriver.exe

as different chrome versions need different drivers and manually installing them is annoying.

```python
all_jobs=[] 
dict={} 
for job in jobs: 
    title=job.find('a',class_='title')
    title_tag=title.text.strip() 
    comp=job.find('a',class_='comp-name mw-25')
    comp_tag=comp.text.strip()
    rating=job.find('a',class_='rating')
    rating_tag=rating.text.strip()
    exp=job.find('span',class_='exp-wrap')
    exp_tag=exp.text.strip()
    #salary=job.find('span',class_='sal-wrap ver-line')
    #salary_tag=salary.text.strip()
    loc=job.find('span',class_='loc-wrap ver-line')
    loc_tag=loc.text.strip()
    description=job.find('span',class_='job-desc ni-job-tuple-icon ni-job-tuple-icon-srp-description')
    description_tag=description.text.strip()
    skills=job.find_all('li',class_="dot-gt tag-li")
    skills_tag=[skill.text.strip() for skill in skills]
    dict={"title":title_tag, "company":comp_tag, "rating":rating_tag, 
          "experience":exp_tag, "location":loc_tag,
            "description":description_tag, "skills":skills_tag} 
    all_jobs.append(dict)
print(all_jobs[0])
```

```python
#pagination
alljobs=[]
for page in range(0,10):
    url=f"https://www.naukri.com/data-scientist-jobs-{page}?k=data%20scientist&nignbevent_src=jobsearchDeskGNB"
    driver.get(url)
    time.sleep(1)
    soup=BeautifulSoup(driver.page_source,'html.parser')
    container=soup.find('div',id='listContainer')
    jobs=container.find_all('div',class_='srp-jobtuple-wrapper')
    for job in jobs: 
        title=job.find('a',class_='title')
        title_tag=title.text.strip() 
        comp=job.find('a',class_='comp-name mw-25')
        comp_tag=comp.text.strip()
        rating=job.find('a',class_='rating')
        rating_tag=rating.text.strip()
        exp=job.find('span',class_='exp-wrap')
        exp_tag=exp.text.strip()
        #salary=job.find('span',class_='sal-wrap ver-line')
        #salary_tag=salary.text.strip()
        loc=job.find('span',class_='loc-wrap ver-line')
        loc_tag=loc.text.strip()
        description=job.find('span',class_='job-desc ni-job-tuple-icon ni-job-tuple-icon-srp-description')
        description_tag=description.text.strip()
        skills=job.find_all('li',class_="dot-gt tag-li")
        skills_tag=[skill.text.strip() for skill in skills]
        dict={"title":title_tag, "company":comp_tag, "rating":rating_tag, 
              "experience":exp_tag, "location":loc_tag,
                "description":description_tag, "skills":skills_tag} 
        alljobs.append(dict)
```
